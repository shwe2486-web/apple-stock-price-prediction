```python
# Step 7
# Moving averages smooth noise and reveal trends.

# We use:

# 30-day MA → Short-term trend
# 90-day MA → Long-term trend

# Create moving averages

df['MA_30'] = df['Close'].rolling(window=10).mean()

df['MA_90'] = df['Close'].rolling(window=50).mean()
```

```python
# df['Close']: Selects the 'Close' column, typically representing closing stock prices.
# .rolling(window=30): Creates a rolling window object. It gathers 30 consecutive data points (the current row and 29 previous rows).
# .mean(): Calculates the average value of those 30 data points in the window.
```

```python
# Daily Returns
df['Daily_Return'] = df['Close'].pct_change()
```

```python
# Volatility
df['Volatility'] = df['Daily_Return'].rolling(window=10).std()
```

```python
# Step 12 — Check Stationarity (ADF Test)
# ARIMA requires stationary data.

# ADF Test checks:

# Whether trend exists
# Whether differencing is needed
```

```python
from statsmodels.tsa.stattools import adfuller

# Perform ADF test

result = adfuller(df['Close'].dropna())

print("ADF Statistic:", result[0])

print("p-value:", result[1])
```

```python
# p-value ≤ 0.05: Reject the null hypothesis; the series is stationary.(desired)
# p-value > 0.05: Fail to reject the null hypothesis; the series is non-stationary.
```

```python
# Step 13 — Feature Engineering
# New features improve prediction accuracy.
```

```python
# Exponential Moving Average

df['EMA_30'] = df['Close'].ewm(span=30).mean()

# Lag features

df['Lag_1'] = df['Close'].shift(1)

df['Lag_2'] = df['Close'].shift(2)

df['Lag_3'] = df['Close'].shift(3)
```

```python
# df['Close']: Selects the 'Close' column (typically stock prices) from the DataFrame df..ewm(span=30): Initializes an Exponentially
# Weighted Moving window object. The span parameter corresponds to an -day EMA, where.It calculates the smoothing factor () 
# automatically using the formula: .mean(): Calculates the mean of the exponentially weighted data.
```
